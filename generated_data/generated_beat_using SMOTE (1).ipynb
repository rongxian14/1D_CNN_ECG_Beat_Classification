```python
import numpy as np
import pickle
import os
from collections import Counter
from sklearn.preprocessing import LabelEncoder
from imblearn.over_sampling import SMOTE

# === Load Data ===
X_all_normalized = np.load('/content/drive/My Drive/ECG_dataset/preprocessed_data/X_all_normalized.npy')
with open('/content/drive/My Drive/ECG_dataset/preprocessed_data/sym_all.pkl', 'rb') as f:
    sym_all = pickle.load(f)

# === Encode Labels ===
ECG_classes = ['N', 'L', 'R', 'V', '/', 'A', 'f', 'F', 'j', 'a', 'E', 'J', 'e', 'S']
label_encoder = LabelEncoder()
label_encoder.classes_ = np.array(ECG_classes)
Y_all = label_encoder.transform(sym_all)

print("Original class distribution:", Counter(Y_all))

# === Target class to resample ===
target_class = 2   # Class 'R'
# target_class = 3   # Class 'V'
# target_class = 4   # Class '/'
# target_class = 5   # Class 'A'
# target_class = 6   # Class 'f'
# target_class = 7   # Class 'F'
# target_class = 8   # Class 'j'
# target_class = 9   # Class 'a'
# target_class = 10  # Class 'E'
# target_class = 11  # Class 'J'
# target_class = 12  # Class 'e'
target_samples = 8000

# === Extract only class 2 and a few others for SMOTE to work ===
# SMOTE needs at least 2 classes to generate samples

mask = (Y_all == target_class)
X_target = X_all_normalized[mask]
Y_target = Y_all[mask]

# Add temporary dummy class to meet SMOTE requirement
X_smote_input = np.vstack([X_target, X_all_normalized[Y_all == 0][:5]])  # Add a few 'N' beats
Y_smote_input = np.hstack([Y_target, Y_all[Y_all == 0][:5]])

# Define SMOTE strategy
smote = SMOTE(sampling_strategy={target_class: target_samples}, random_state=42, k_neighbors=5)
X_resampled, Y_resampled = smote.fit_resample(X_smote_input, Y_smote_input)

# Filter only the resampled class 2 beats (ignore dummy class)
mask_resampled = (Y_resampled == target_class)
X_class2_resampled = X_resampled[mask_resampled]
Y_class2_resampled = Y_resampled[mask_resampled]

print(f"✅ After SMOTE: Class 2 ('R') beat count = {X_class2_resampled.shape[0]}")

# === Save the resampled class 2 beats ===
save_path = '/content/drive/My Drive/ECG_dataset/SMOTEresampling/resampled_class_2'
os.makedirs(save_path, exist_ok=True)

np.save(os.path.join(save_path, 'X_generate_2.npy'), X_class2_resampled)
np.save(os.path.join(save_path, 'Y_generate_2.npy'), Y_class2_resampled)

print(f"✅ Saved to {save_path}/")
```
